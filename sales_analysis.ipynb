```python
X1 = ["#A94C1A", "#ED944C"]

plt.figure(figsize=(15,6))

A1 = sns.countplot(x="Zone",data=df,hue="Gender",palette=X1,edgecolor="black")

for i in A1.containers:
    A1.bar_label(i)

plt.title("Gender vs Zone Wise Count",color="#60280A",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
C4 = ["#00695C", "#008577", "#2A9D8F", "#52B6A8", "#80CBC4", "#B2DFDB"]

piechart = df.groupby("Zone",as_index=False)["User_ID"].count()

plt.figure(figsize=(15,6))

plt.pie(piechart["User_ID"],labels=piechart["Zone"],autopct="%.f%%",startangle=110,colors=C4)

plt.title("Zone Wise Total Users",color="#11564E",fontweight="bold", fontsize=16, family="sans-serif")

plt.show()
```

```python
plt.figure(figsize=(15,6))

C7 = ["#4B0082", "#8A3EE1"]

A2 = sns.countplot(x='Age Group',data=df,hue='Marital_Status',palette=C7,edgecolor="black")

for i in A2.containers:
    A2.bar_label(i)

plt.title("Marital Status Wise Age Group Count",color="#47116D",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python


Linechart = df.groupby("Sector",as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)

Linechart
```

```python
plt.figure(figsize=(15,6))

sns.lineplot(x="Sector",y="Amount",data=Linechart,marker="o")

plt.xticks(rotation=90)

plt.grid()

plt.title("Sector Wise Total Amount in Sorted Format")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()
```

```python
chart_2 = df.groupby("State",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
```

```python
plt.figure(figsize=(15,6))

sns.barplot(x="State",y="Orders",data=chart_2,palette="tab20")

plt.xticks(rotation=90)

plt.title("Top 10 State Order Wise",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()

plt.show()
```
